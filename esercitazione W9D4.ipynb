# Esercizio 1 · Costrutto if/elif/else (semplice · 20 minuti)

## Esercizio 1.1
**Domanda:** con `temperatura = 78` scrivere un costrutto if/elif/else che stampi "Allarme" (sopra 80), "Attenzione" (sopra 60) o "Normale" (altrimenti), usando gli operatori di confronto corretti.

```python
temperatura = 78
print("Temperatura:", temperatura, type(temperatura))

if temperatura > 80:
    print("Allarme")
elif temperatura > 60:
    print("Attenzione")
else:
    print("Normale")
```

## Esercizio 1.2
**Domanda:** modificare il valore di temperatura con altri due valori a scelta, uno che attivi il ramo `elif` e uno che attivi il ramo `else`, eseguendo di nuovo il codice per ciascuno.

```python
# Valore che attiva il ramo elif (60 < t <= 80)
temperatura = 70
print("Temperatura:", temperatura)

if temperatura > 80:
    print("Allarme")
elif temperatura > 60:
    print("Attenzione")
else:
    print("Normale")
```

```python
# Valore che attiva il ramo else (t <= 60)
temperatura = 45
print("Temperatura:", temperatura)

if temperatura > 80:
    print("Allarme")
elif temperatura > 60:
    print("Attenzione")
else:
    print("Normale")
```

```python
# Verifica aggiuntiva: valore che attiva il ramo if (t > 80)
temperatura = 85
print("Temperatura:", temperatura)

if temperatura > 80:
    print("Allarme")
elif temperatura > 60:
    print("Attenzione")
else:
    print("Normale")
```

## Esercizio 1.3
**Domanda:** spiegare, in una frase, perché in nessun caso vengono eseguiti più di uno tra i rami if, elif ed else.

**Risposta:** Python valuta le condizioni dall'alto verso il basso e, appena una risulta `True`, esegue solo quel blocco e salta tutti i rami successivi; l'`else` viene eseguito solo se tutte le condizioni precedenti sono `False`.

### Definizione richiesta
**Costrutto condizionale:** struttura che esegue un solo blocco di codice tra più alternative, in base al valore (`True`/`False`) di una condizione booleana.

**Differenza tra elif ed else:** `elif` ha una propria condizione da verificare e viene valutato solo se le condizioni precedenti sono false; `else` non ha condizione e cattura tutti i casi rimasti.

# Esercizio 2 · Ciclo for (semplice · 20 minuti)

## Esercizio 2.1
**Domanda:** usare un ciclo for per stampare ogni elemento della lista `articoli`, uno per riga.

```python
articoli = ["penna", "quaderno", "zaino"]
prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}
print(type(articoli), type(prezzi))

for articolo in articoli:
    print(articolo)
```

## Esercizio 2.2
**Domanda:** usare un ciclo for con il metodo `.items()` per stampare ogni coppia articolo-prezzo del dizionario `prezzi`.

```python
for articolo, prezzo in prezzi.items():
    print(articolo, "->", prezzo, "€")
```

## Esercizio 2.3
**Domanda:** all'interno del ciclo sul dizionario, usare un if per stampare un messaggio aggiuntivo solo per gli articoli con prezzo superiore a 10.

```python
soglia = 10

for articolo, prezzo in prezzi.items():
    print(articolo, "->", prezzo, "€")
    if prezzo > soglia:
        print("   ⚠ Articolo sopra soglia:", articolo, "costa più di", soglia, "€")
```

### Definizione richiesta
**Ciclo for:** costrutto che itera su un iterabile (lista, tupla, set, stringa, dizionario) eseguendo lo stesso blocco di codice per ciascun elemento.

**Differenza tra iterare su un dizionario e iterare con `.items()`:** iterando direttamente sul dizionario si ottengono solo le **chiavi**; con `.items()` si ottiene a ogni giro la **coppia (chiave, valore)**, come mostra la cella seguente.

```python
print("Iterazione diretta sul dizionario:")
for elemento in prezzi:
    print(" ", elemento)

print("Iterazione con .items():")
for elemento in prezzi.items():
    print(" ", elemento)
```

# Esercizio 3 · Funzioni con parametri e return (semplice · 20 minuti)

## Esercizio 3.1
**Domanda:** definire una funzione `calcola_totale(prezzo, quantita)` che restituisca con `return` il prodotto tra i due parametri.

```python
def calcola_totale(prezzo, quantita):
    return prezzo * quantita

print(type(calcola_totale))
```

## Esercizio 3.2
**Domanda:** richiamare la funzione una volta con argomenti posizionali e una volta con argomenti keyword invertendo l'ordine, e stampare entrambi i risultati.

```python
totale_posizionale = calcola_totale(45.90, 3)
totale_keyword = calcola_totale(quantita=3, prezzo=45.90)

print("Argomenti posizionali:", round(totale_posizionale, 2))
print("Argomenti keyword (ordine invertito):", round(totale_keyword, 2))
```

## Esercizio 3.3
**Domanda:** modificare la funzione aggiungendo un parametro `sconto` con valore di default 0, che riduce il totale della percentuale indicata quando viene passato un valore diverso da 0.

```python
def calcola_totale(prezzo, quantita, sconto=0):
    totale = prezzo * quantita
    if sconto != 0:
        totale = totale * (1 - sconto / 100)
    return round(totale, 2)

print("Posizionale, senza sconto:", calcola_totale(45.90, 3))
print("Keyword, senza sconto:   ", calcola_totale(quantita=3, prezzo=45.90))
print("Con sconto del 10%:      ", calcola_totale(45.90, 3, sconto=10))
```

**Nota:** una funzione senza `return` restituisce `None`; appena Python esegue `return`, la funzione termina e le righe successive nello stesso blocco non vengono eseguite.

### Definizione richiesta
**Funzione:** blocco di codice riutilizzabile, definito con `def`, che riceve dei parametri e può restituire un valore con `return`; viene eseguito solo quando è richiamato con il suo nome seguito da parentesi.

**Parametro con valore di default:** parametro a cui è assegnato un valore nella definizione (`sconto=0`), usato automaticamente se nella chiamata non viene passato nessun argomento. L'argomento `sconto` può essere omesso **quando non si vuole applicare alcuno sconto**, perché in quel caso vale già 0.

# Esercizio 4 · Ciclo while (avanzato · 30 minuti)

## Esercizio 4.1 e 4.2
**Domanda:** scrivere un ciclo while che aggiorni un capitale iniziale di 1000 € (crescita del 5% annuo) finché non supera 1500 €, aggiornando a ogni iterazione sia il capitale sia il contatore degli anni, e stampare capitale e anni trascorsi a ogni iterazione.

```python
capitale = 1000.0
tasso = 0.05
obiettivo = 1500
anni = 0
print(type(capitale), type(anni))

while capitale <= obiettivo:
    capitale = capitale * (1 + tasso)   # istruzione critica: aggiorna la condizione
    anni += 1
    print(f"Anno {anni}: capitale = {capitale:.2f} €")

print()
print("Anni necessari per superare", obiettivo, "€:", anni)
```

## Esercizio 4.3
**Domanda:** indicare quale istruzione, se rimossa dal corpo del ciclo, trasformerebbe questo while in un loop infinito, e per quale motivo.

**Risposta:** l'istruzione critica è `capitale = capitale * (1 + tasso)`: è l'unica che modifica la variabile usata nella condizione `capitale <= obiettivo`; senza di essa il capitale resterebbe sempre 1000, la condizione sarebbe sempre `True` e il ciclo non terminerebbe mai. Rimuovere `anni += 1` invece darebbe un conteggio sbagliato, ma il ciclo finirebbe comunque.

### Definizione richiesta
**Ciclo while:** costrutto che ripete un blocco di codice finché una condizione resta vera.

**Differenza rispetto al for:** il `for` si usa quando il numero di iterazioni è **noto in anticipo** (un giro per ogni elemento dell'iterabile); il `while` si usa quando **non è noto in anticipo** e dipende da una condizione, come qui, dove gli anni necessari si scoprono solo durante il calcolo.

# Esercizio 5 · List comprehension (avanzato · 30 minuti)

## Esercizio 5.1
**Domanda:** scrivere un ciclo for con `append()` che costruisca una nuova lista contenente solo i voti maggiori o uguali a 6.

```python
voti = [5, 8, 6, 9, 4, 7, 10]
print(type(voti))

sufficienti_for = []
for voto in voti:
    if voto >= 6:
        sufficienti_for.append(voto)

print("Con ciclo for:", sufficienti_for)
```

## Esercizio 5.2
**Domanda:** scrivere una list comprehension che ottenga lo stesso risultato del punto 1, con la sintassi `[espressione for elemento in iterabile if condizione]`.

```python
sufficienti_comp = [voto for voto in voti if voto >= 6]

print("Con list comprehension:", sufficienti_comp)
print("I due risultati coincidono?", sufficienti_for == sufficienti_comp)
```

## Esercizio 5.3
**Domanda:** scrivere una seconda list comprehension che moltiplichi per 10 ciascun voto della lista, senza alcuna condizione di filtro.

```python
voti_per_10 = [voto * 10 for voto in voti]

print("Voti x 10:", voti_per_10)
print("Lunghezza voti:", len(voti), "| Lunghezza voti_per_10:", len(voti_per_10))
```

### Definizione richiesta
**List comprehension:** sintassi compatta che crea una nuova lista applicando un'espressione a ogni elemento di un iterabile, con un filtro opzionale.

**Quando è preferibile tornare a un ciclo for:**
- quando la logica richiede più istruzioni o condizioni annidate (if/elif/else complessi);
- quando non serve costruire una lista ma eseguire azioni (stampare, salvare, aggiornare variabili);
- quando la comprehension diventerebbe troppo lunga e quindi poco leggibile;
- quando serve interrompere il ciclo con `break` o gestire errori con `try/except`.

# Esercizio 6 · Funzioni e librerie (avanzato · 30 minuti)

## Esercizio 6.1
**Domanda:** importare la libreria `math` e definire una funzione `radice_arrotondata(numero, decimali=2)` che restituisca con `return` la radice quadrata di numero, arrotondata al numero di decimali indicato con `round()`.

```python
import math

def radice_arrotondata(numero, decimali=2):
    return round(math.sqrt(numero), decimali)

print(type(math), type(radice_arrotondata))
```

## Esercizio 6.2
**Domanda:** richiamare la funzione su ciascuno dei 5 numeri usando un ciclo for, stampando ogni risultato.

```python
numeri = [2, 10, 16, 50, 123]

for numero in numeri:
    print(f"Radice di {numero}: {radice_arrotondata(numero)}")
```

## Esercizio 6.3
**Domanda:** richiamare la funzione una volta senza indicare l'argomento `decimali` e una volta indicandolo esplicitamente come argomento keyword con un valore diverso da 2, confrontando i due output.

```python
con_default = radice_arrotondata(2)
con_keyword = radice_arrotondata(2, decimali=5)

print("Default (2 decimali): ", con_default)
print("Keyword decimali=5:   ", con_keyword)
```

### Definizione richiesta
**Libreria:** raccolta di moduli con funzioni, classi e costanti già pronte da riutilizzare nel proprio codice (ad esempio `math`, `random`, `datetime`).

**Perché serve `import` anche per la standard library:** i moduli della standard library sono installati insieme a Python ma **non vengono caricati automaticamente** in memoria; finché non si scrive `import math`, il nome `math` non esiste nel programma e una chiamata a `math.sqrt()` genera un `NameError`. Solo le funzioni *built-in* (`print`, `len`, `round`…) sono sempre disponibili.
